```python



from json import tool
from os import name

from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain.messages import HumanMessage,AIMessage,SystemMessage
from vector import similarity_search


sys_msg = SystemMessage(content="You are a helpfull AI Youtube Chat Bot You can Ansewer according to the video Transript Context ",name = "Saroj.ai")

load_dotenv()
message = [sys_msg]




llm = ChatGroq(
    model='openai/gpt-oss-120b'
)
Active  = True
while Active == True:

    User = input("Enter your message")
    user_msg = HumanMessage(User)
    if User == "exit":
        Active = False
    message.append(user_msg)
    respose  = llm.invoke(message)
    ai_msg = AIMessage(respose.content)
    print(ai_msg.content)
    message.append(ai_msg)







```
